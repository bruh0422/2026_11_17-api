And run the following code to define the `to_json` function, which will be used to save the API response data to a JSON file:

````python
import json

def to_json(data, filename: str) -> None:
    with open(filename, 'w') as f:
        json.dump(data, f, ensure_ascii=False, indent=4)

print('The `to_json` function has been defined. You can use it to save API response data to a JSON file by calling to_json(data, filename).')
````

## Bevy
https://help.bevy.com/hc/en-us/articles/21156148264215-Bevy-API-Reference

````python
import requests # 引入 requests 模組

url = 'https://gdg.community.dev/api/chapter/1663/event/' # Base URL & Endpoint

response = requests.request('GET', url) # GET 方法請求 API

data = response.json() # 將回傳的 JSON 轉換為 Python 物件

print(data)                # 印出回傳的資料
to_json(data, 'bevy.json') # 將資料寫入 bevy.json
````

### Extract Data

````python
print(data['results'])             # 所有活動
print(data['results'][0])          # 最新活動
print(data['results'][0]['title']) # 最新活動標題

print('==============================')

for event in data['results'][:5]:  # 前 5 個活動
    print(event['title'])          # 列出活動標題
````

### Parse & Extract Data
找到對應的資訊並填入下方程式的 `...` 中！

````python
filtered_data = {}

for station in data['Data']['retVal']:             # 遍歷所有站點資料
    if ... in station[...]:                        # 如果站點名稱 (sna) 包含「中山大學」
        filtered_data[station[...]] = {            # 將站點名稱 (sna) 設為 key 寫入
            'avaliable': {
                'regular': int(station[...][...]), # 一般車輛 (sbi_detail / yb2)
                'ebike': int(station[...][...])    # 電輔車 (sbi_detail / eyb)
            },
            'empty_docks': int(station[...]),      # 可借 (bemp)
            'is_active': int(station[...])         # 站點是否啟用 (act)
        }

print(filtered_data)                          # 印出篩選後的資料
to_json(filtered_data, 'ubike_filtered.json') # 將篩選後的資料寫入 ubike_filtered.json
````
